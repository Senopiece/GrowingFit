# Обычная MLP: 128 ReLU-нейронов

$$f(x)=\sin(1.7x)+0.35\sin(5.3x)+0.15x^2-0.3x, \qquad x\in[-3,3].$$

Один скрытый слой сразу на 128 нейронов. Все нейроны и выходной bias обучаются совместно с первого шага, без роста и заморозки во время обучения. 1600 шагов Adam с MSE. Обучение в фоне без пауз, отрисовка по запросу браузера с ограничением `target_fps`.

Лосс сверху, целевая функция и текущая аппроксимация снизу. Ноутбук запускается самостоятельно, общий код находится в `_shared/`.

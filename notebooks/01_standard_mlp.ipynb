# Standard MLP

$$f(x)=\sin(1.7x)+0.35\sin(5.3x)+0.15x^2-0.3x, \qquad x\in[-3,3].$$

Architecture: `Linear(1, hidden_size) → ReLU → Linear(hidden_size, 1)`. All parameters are trained jointly using Adam and standard MSE. `hidden_size` sets the number of hidden neurons; `total_steps` sets the number of updates.

Training runs in a background thread. The renderer requests the latest model, evaluates MSE and the fit, and updates Plotly with a `target_fps` limit. Loss is shown above the function approximation; frames are never queued.

```python
from pathlib import Path
import sys

project_root = next(
    p
    for p in (Path.cwd(), *Path.cwd().parents)
    if (p / "pyproject.toml").is_file() and (p / "notebooks" / "_shared").is_dir()
)
shared_parent = str(project_root / "notebooks")
if shared_parent not in sys.path:
    sys.path.insert(0, shared_parent)

import torch
from _shared import Config, Experiment
from _shared.visualization import train_live

torch.set_num_threads(1)
```

```python
config = Config(
    seed=42,
    hidden_size=32,
    total_steps=32_000,
    learning_rate=0.01,
    n_train=256,
    n_grid=601,
    x_min=-3.0,
    x_max=3.0,
)
target_fps = 10
```

```python
baseline_experiment = Experiment(config)
await train_live(baseline_experiment, target_fps=target_fps)
```
